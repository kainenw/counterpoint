```python
import numpy as np
import sounddevice as sd

def play(frequencies, duration):
  # Generate the time axis
  t = np.linspace(0, duration, int(duration * 44100), endpoint=False)
  
  # Generate the combined sine wave
  sine_wave = np.sum([np.sin(2 * np.pi * f * t) for f in frequencies], axis=0)
  
  # Play the sine wave
  sd.play(sine_wave, samplerate=44100)
  sd.wait()

# Define the parameters
duration = .1  # Duration of the sine wave in seconds
frequencies = [220, 330, 440, 550, 660, 770, 880]  # Frequencies of the sine waves in Hz

play([440, 660], duration)
play([550, 770], duration)
play([660, 880], duration)
play([770, 990], duration)
play([880, 1100], duration)
play([770, 990], duration)
play([660, 880], duration)
play([550, 770], duration)
play([440, 660], duration)


```
