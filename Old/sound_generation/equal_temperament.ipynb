```python
import math

def ET(octave_divisions):
    # ratio between semitones
    semitone = math.pow(2, 1 / octave_divisions)

    def hz(num):
        ref_hz = 440  # (A4 = 440Hz) reference pitch
        ref_num = 49  # (A4) reference number
        # ratio from base frequency
        distance = num - ref_num
        
        ratio = math.pow(semitone, distance)

        # return frequency
        return ref_hz * ratio

    tones = list()
    for i in range(0, 88):
        pitch = hz(i)
        tones.append(pitch)

    return tones
```

```python
def convert_to_frequencies(numbers, key_center):
  TTET = ET(12)
  frequencies = []
  for num in numbers:
    frequency = TTET[num + key_center]
    frequencies.append(frequency)
  return frequencies

notes = [0, 2, 4, 5, 7, 9, 11]
expected = [440.0, 493.8833012561241, 554.3652619537443, 587.3295358348153, 659.2551138257401, 739.988845423269, 830.6093951598907]
print(convert_to_frequencies(notes, 49) == expected)
```
